```python
import numpy as np
```

```python
df.loc[
    df['Review_Rating'] > 5,
    'Review_Rating'
] = np.nan
```

```python
rating_medians = (
    df.loc[
        (df['Purchased'] == 'Yes') &
        (df['Review_Rating'].notna())
    ]
    .groupby('Product_Category')['Review_Rating']
    .median()
)

rating_medians
```

```python
df.loc[
    (df['Purchased'] == 'Yes') & (df['Review_Rating'].isna()),
    'Review_Rating'
] = (
    df.loc[
        (df['Purchased'] == 'Yes') & (df['Review_Rating'].isna()),
        'Product_Category'
    ].map(rating_medians)
)
```

```python
df['Review_Rating'].value_counts(dropna=False).sort_index()
```

```python
satisfaction_medians = (
    df.loc[
        (df['Purchased'] == 'Yes') &
        (df['Customer_Satisfaction'].notna())
    ]
    .groupby('Product_Category')['Customer_Satisfaction']
    .median()
)

satisfaction_medians
```

```python
df.loc[
    (df['Purchased'] == 'Yes') & (df['Customer_Satisfaction'].isna()),
    'Customer_Satisfaction'
] = (
    df.loc[
        (df['Purchased'] == 'Yes') & (df['Customer_Satisfaction'].isna()),
        'Product_Category'
    ].map(satisfaction_medians)
)
```
